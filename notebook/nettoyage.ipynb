##  Dispo Alimentaire

```python
# 1) Charger
fbs_raw = pd.read_csv("../data_raw/dispo_alimentaire.csv")

# Harmoniser maïs
fbs_raw["Item"] = fbs_raw["Item"].replace({
    "Maize": "Maize and products"
})

# 2) Produits à garder
items_keep = [
    "Wheat and products",
    "Rice (milled equivalent)",
    "Maize and products"
]

# 3) Éléments à garder
elements_keep = {
    "Production": "production",
    "Import quantity": "import",
    "Export quantity": "export",
    "Food supply (kg/capita/yr)": "supply_kg_ph"
}

# 4) Filtrer
fbs = fbs_raw[
    fbs_raw["Item"].isin(items_keep) &
    fbs_raw["Element"].isin(elements_keep.keys())
][["Area", "Year", "Item", "Element", "Value"]]

# 🔥 Garder uniquement 2022
if 2022 in fbs["Year"].unique():
    fbs = fbs[fbs["Year"] == 2022]
else:
    print("⚠️ Pas d'année 2022 dans ce fichier.")

# 5) Renommer les éléments
fbs["Element_simple"] = fbs["Element"].map(elements_keep)

# 6) Créer un nom de variable par ligne
fbs["varname"] = (
    fbs["Item"].str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "")
    .str.replace(")", "")
    .str.replace("-", "_")
    .str.replace("/", "_")
    + "_" +
    fbs["Element_simple"]
)

# 7) Pivot
fbs_simple_pivot = fbs.pivot_table(
    index=["Area"],      # 👈 plus de Year ici
    columns="varname",
    values="Value",
    aggfunc="first"
).reset_index()

# 8) Agrégation
fbs_simple_pivot["cereal_production_t"] = fbs_simple_pivot.filter(regex="production$").sum(axis=1)
fbs_simple_pivot["cereal_import_t"]     = fbs_simple_pivot.filter(regex="import$").sum(axis=1)
fbs_simple_pivot["cereal_export_t"]     = fbs_simple_pivot.filter(regex="export$").sum(axis=1)
fbs_simple_pivot["cereal_supply_kg_ph"] = fbs_simple_pivot.filter(regex="supply_kg_ph$").sum(axis=1)

# 9) Finaliser (🔥 suppression de Year)
fbs_cereal_final = fbs_simple_pivot[[
    "Area",
    "cereal_production_t",
    "cereal_import_t",
    "cereal_export_t",
    "cereal_supply_kg_ph"
]].rename(columns={"Area": "Country"})

# 10) Convertir en numérique
numeric_cols = [
    "cereal_production_t",
    "cereal_import_t",
    "cereal_export_t",
    "cereal_supply_kg_ph"
]

for col in numeric_cols:
    fbs_cereal_final[col] = pd.to_numeric(fbs_cereal_final[col], errors="coerce")

fbs_cereal_final.head()
```

## Vérifier les pays et les Annéees

```python
# Chargement des données brutes 

gdp_raw    = pd.read_csv("../data_raw\API_NY.GDP.PCAP.PP.CD_DS2_en_csv_v2_129964.csv", skiprows=4)
log_raw    = pd.read_csv("../data_raw/API_LP.LPI.OVRL.XQ_DS2_en_csv_v2_6417.csv", skiprows=4)
stab_raw   = pd.read_csv("../data_raw\API_PV.EST_DS2_en_csv_v2_126909.csv", skiprows=4)
tariff_raw = pd.read_csv("../data_raw\API_TM.TAX.MRCH.SM.AR.ZS_DS2_en_csv_v2_126349.csv", skiprows=4)
urban_raw  = pd.read_csv("../data_raw/API_SP.URB.TOTL.IN.ZS_DS2_en_csv_v2_129596.csv", skiprows=4)
pop_raw    = pd.read_csv("../data_raw/API_SP.POP.TOTL_DS2_en_csv_v2_130083.csv", skiprows=4)
menage_raw   = pd.read_csv("../data_raw\API_NE.CON.PRVT.PC.KD_DS2_en_csv_v2_6439.csv", skiprows=4)    

# Fonction générique pour transformer les fichiers World Bank du format large (colonnes = années)
# en format long (année en ligne). var_name = nom de variable (GDP, stabilité, etc.)
def wb_long(df, var_name):
    year_cols = [c for c in df.columns if c.isdigit()]
    # Sélection des colonnes correspondant aux années (colonnes numériques)
    # Transformation wide → long
    df_long = df.melt(
        id_vars=["Country Name", "Country Code"],
        value_vars=year_cols,
        var_name="Year",
        value_name=var_name
    )
    # Conversion correcte des années
    df_long["Year"] = df_long["Year"].astype(int)
    # Renommage pour harmonisation
    df_long = df_long.rename(columns={
        "Country Name": "Country",
        "Country Code": "ISO3"
    })
    return df_long

# Conversion de toutes les sources WB au format long uniforme
gdp_long    = wb_long(gdp_raw,    "gdp_ppp_pc")
log_long    = wb_long(log_raw,    "logistics_index")
stab_long   = wb_long(stab_raw,   "political_stability")
tariff_long = wb_long(tariff_raw, "tariff_rate")
urban_long  = wb_long(urban_raw,  "urban_rate")
pop_long    = wb_long(pop_raw,    "population")
```

```python
# Chargement des données FAO & OWID (déjà en format long)
fbs_raw      = pd.read_csv("../data_raw/food_balances_poultry.csv")      # Area, Year
livestock_raw= pd.read_csv("../data_raw/livestock_production.csv")      # Area, Year
trade_raw    = pd.read_csv("../data_raw/trade.csv")                     # Area, Year
prices_raw   = pd.read_csv("../data_raw/prices.csv")                    # Area, Year
                

organic_raw  = pd.read_csv("../data_raw/share-of-arable-land-which-is-organic.csv")  # Entity, Year
n2o_raw      = pd.read_csv("../data_raw/per-capita-nitrous-oxide-agriculture.csv")   # Entity, Year

fbs_cereal_df = fbs_cereal_final.copy()

```

```python
# Assemblage dans un dictionnaire pour automatiser l'analyse de couverture
# Chaque entrée contient au minimum : colonne pays + année
dfs = {
    "FAO_food_balances": fbs_raw[["Area", "Year"]],
    "FAO_livestock":     livestock_raw[["Area", "Year"]],
    "FAO_trade":         trade_raw[["Area", "Year"]],
    "FAO_prices":        prices_raw[["Area", "Year"]],
    

    "FAO_cereals":       fbs_cereal_final[["Country"]],  

    "OWID_organic":      organic_raw[["Entity", "Year"]],
    "OWID_n2o":          n2o_raw[["Entity", "Year"]],

    "WB_gdp":            gdp_long[["Country", "Year"]],
    "WB_logistics":      log_long[["Country", "Year"]],
    "WB_stability":      stab_long[["Country", "Year"]],
    "WB_tariff":         tariff_long[["Country", "Year"]],
    "WB_urban":          urban_long[["Country", "Year"]],
    "WB_population":     pop_long[["Country", "Year"]],
}

# Fonction pour calculer, pour chaque année, le nombre de pays présents dans une source
def coverage(df, name):
    # deviner la colonne pays
    if "Country" in df.columns:
        country_col = "Country"
    elif "Area" in df.columns:
        country_col = "Area"
    elif "Entity" in df.columns:
        country_col = "Entity"
    elif "Enquête" in df.columns:
        country_col = "Enquête"   # on comptera les enquêtes (1 par pays/année)
    else:
        print(f"{name}: pas de colonne pays reconnue")
        return pd.DataFrame()
    # Calcul : nb de pays uniques par année
    cov = (
        df.groupby("Year")[country_col]
        .nunique()
        .reset_index()
        .rename(columns={country_col: "nb_pays"})
    )
    cov["source"] = name
    return cov
```

```python
# Calcul de la couverture pour toutes les sources
all_cov = []

for name, df in dfs.items():
    try:
        cov = coverage(df, name)
        all_cov.append(cov)
    except Exception as e:
        print(f"⚠️ Erreur pour {name}: {e}")

coverage_all = pd.concat(all_cov, ignore_index=True)

# Résumé : années min/max couvertes par chaque source
print(
    coverage_all
    .groupby("source")["Year"]
    .agg(["min", "max"])
    .sort_values("max")
)
```

```python
# Moyenne du nombre de pays couverts par année
summary = (
    coverage_all
    .groupby("Year")["nb_pays"]
    .mean()
    .reset_index()
    .sort_values("Year", ascending=False)
)

summary.head(15)
```

```python
# Tableau pivot année × source → nb de pays
pivot = (
    coverage_all
    .pivot(index="Year", columns="source", values="nb_pays")
    .sort_index(ascending=False)
)

pivot.head(10)

# Indicateurs supplémentaires : nombre de sources avec <100 pays
pivot["nb_sources_sous_100"] = (pivot < 100).sum(axis=1)
pivot["nb_sources_total"] = pivot.notna().sum(axis=1)

pivot["pct_sources_sous_100"] = pivot["nb_sources_sous_100"] / pivot["nb_sources_total"] * 100

# Fusion avec la moyenne du nombre de pays
summary = (
    coverage_all
    .groupby("Year")["nb_pays"]
    .mean()
    .rename("moyenne_nb_pays")
)

pivot = pivot.merge(summary, on="Year")

pivot.head(10)
```

```python
def keep_real_countries(df):
    # Garder seulement les ISO3 non-na
    df = df.dropna(subset=["ISO3"]).copy()
    # Garder uniquement les codes ISO3 valides (3 lettres A-Z)
    df = df[df["ISO3"].str.fullmatch(r"[A-Z]{3}")]
    return df

logistique_2022 = keep_real_countries(logistique_2022)
pib_2022 = keep_real_countries(pib_2022)
stabilite_2022 = keep_real_countries(stabilite_2022)
pop_2022 = keep_real_countries(pop_2022)
urban_2022 = keep_real_countries(urban_2022)
douane_2022 = keep_real_countries(douane_2022)
n2o_2022 = keep_real_countries(n2o_2022)
organic_2022 = keep_real_countries(organic_2022)
menage_2022 = keep_real_countries(menage_2022)
```

```python
#Problème sur les ISO3 et correction
df_BAO["ISO3"].apply(type).unique()
df_FAO_iso["ISO3"].apply(type).unique()

def fix_iso3(x):
    if isinstance(x, list) and len(x) > 0:
        return x[0]        # garde la première valeur de la liste
    else:
        return x           # garde la valeur telle quelle si c'est déjà une string

df_FAO_iso["ISO3"] = df_FAO_iso["ISO3"].apply(fix_iso3)
df_BAO["ISO3"] = df_BAO["ISO3"].apply(fix_iso3)

```

```python
df_merged = df_BAO.merge(
    df_FAO_iso,
    on="ISO3",
    how="left"   # left pour garder tous les pays BAO
)

df_merged.head()
```

```python
print("Avant :", df_BAO.shape)
print("Après :", df_merged.shape)
```

```python
# Calcul du taux de remplissage pour chaque ligne
df_merged["taux_remplissage"] = df_merged.notna().mean(axis=1) * 100

# Arrondir pour lisibilité
df_merged["taux_remplissage"] = df_merged["taux_remplissage"].round(1)

df_merged.head()
```

```python
nb_pays_80 = (df_merged["taux_remplissage"] > 80).sum()
print("Nombre de pays avec un taux de remplissage > 80% :", nb_pays_80)
```

```python
# Colonnes que tu veux mettre en premier
cols_first = ["Country_x", "taux_remplissage","nb_colonnes_remplies"]

# Colonnes restantes du DataFrame
cols_rest = [c for c in df_merged.columns if c not in cols_first]

# Ajouter la colonne nombre_colonnes_remplies
df_merged["nb_colonnes_remplies"] = df_merged.notna().sum(axis=1)

# Filtrer et réordonner
pays_80 = df_merged[df_merged["taux_remplissage"] > 80][
    cols_first + cols_rest
].sort_values("taux_remplissage", ascending=False).reset_index(drop=True)

pays_80.head()

```
